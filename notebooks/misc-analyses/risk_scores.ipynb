# Overview

Develop a score with the xenium data pseudobulked melanoma cells. 

**Score 1: AM/CM** 
A multiplicative expression of 10 genes, divided by multiplicative expression of the other 10 genes, designed to classify Acral melanoma (AM): Cutaneous melanoma (CM) cells. See: `references/Deacon MRAYIA Project Description.pdf` figure 1b.
* AM genes: PDLIM7, TBX4, IGF4, PDLIM3, ID1, MEG3, HOXA13, EGFL7, ID3, HAND2
* CM genes: SNRPD3, WWP2, CALM1, HOXB6, SLITRK2, SRP14, COX4I1, LGI3, APCDD1, MCOLN3
